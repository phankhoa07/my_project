```python
def zigzag_numbers(n) :
    cot = []
    for i in range(n) :
        hang = []
        for j in range(n) :
            so_nhap_vao = int(input())
            if i % 2 == 0 :
                hang.append(so_nhap_vao)
            else:
                hang.insert(0,so_nhap_vao)
        cot.append(hang)
    for i in cot:
        print(i)
        print()
    return 0

n = int(input())
print(zigzag_numbers(n))


                
```
