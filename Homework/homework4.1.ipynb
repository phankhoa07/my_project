# uoc chung a b

```python
a, b = map(int, input().split())
for  i in range(1, a ) :
    if a % i == 0 :
        for u in range(1, b) :
            if b % u == 0 :
                if u  == i :
                    print(u)
```

```python
a, b = map(int, input().split())
if a < b :
    for i in range(1, a) :
        if a % i == 0 :
            if b % i == 0 :
                print(i)
else :
    for i in range(1, b) :
        if b % i == 0 :
            if a % i == 0 :
                print(i)
```

# fibonuci

```python

A = int(input())
so_truoc = 1
so_sau = 1

while so_sau <= A :
    so_fibo = so_sau + so_truoc
    so_truoc = so_sau 
    so_sau = so_fibo
print(so_truoc)    

```
