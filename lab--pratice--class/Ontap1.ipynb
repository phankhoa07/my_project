```python
def smart_taxi_fare(car_type, start_hour, km , wait_minutes) :
    total = 0

    if car_type == "xe 4 cho" :
        if km <= 0.8 :
            total += 11000
        elif km <= 20 :
            total += 13000*km
        else:
            total += 10500*km
    if car_type == "xe 7 cho" :
        if km <= 0.8 :
            total += 12000
        elif km <= 20 :
            total += 14000*km
        else:
            total += 12000*km
         

    count_wait_minutes = wait_minutes // 5  
    if count_wait_minutes == 0 :
        pass
    else:
        total = total + count_wait_minutes*2000

    
    if 0 <= start_hour <= 5 and 22 <= start_hour <= 23 :
        total = total*1,2
    else:
        if km > 30 :
            total = total - total*0.1
    if total < 50000 and km > 5 :
        total = 50000 
    return total

        

        


    
    





car_type = input()
start_hour = int(input())
km = int(input())
wait_minutes = int(input())

print(smart_taxi_fare(car_type, start_hour, km , wait_minutes))





    

```
