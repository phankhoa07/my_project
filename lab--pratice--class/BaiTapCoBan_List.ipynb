### Bài 1: Tìm số lớn nhất của 1 list

```python
n = list(map(int, input().split()))
max = 0
for i in range(len(n)):
    if n[i] > max :
        max = n[i]
print(max)
```

### Bài 7: Kiểm tra một list số nguyên có phải là dãy tăng, dãy giảm hay vô danh
1 3 5 6 7 8 9: Tang
9 8 7 6 5 4 0: Giam
1 2 3 7 1 0 2: Vo danh
1 2 2 4 5 7: Vo danh

```python
n = list(map(int, input().split()))

if len(n) < 2:
    print("vo danh")
else:
    if (n[1] - n[0]) > 0:
        check = 1
    elif (n[1] - n[0]) < 0:
        check = -1
    else:
        check = 0

    for i in range(len(n) - 2):
        if (n[i + 2] - n[i + 1]) > 0:
            check += 1
        elif (n[i + 2] - n[i + 1]) < 0:
            check -= 1

    if check == len(n) - 1:
        print("day tang")
    elif check == -(len(n) - 1):
        print("day giam")
    else:
        print("vo danh")
```
