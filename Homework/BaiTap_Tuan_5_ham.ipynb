```python
def tong_cac_chu_so(n) :
    total = 0
    while n != 0 :
        total = total + (n%10)
        n = n //10
    return total
n = int(input())
print(tong_cac_chu_so(n))
    
```

```python
def prime_number_check ( n ) :
    if n < 2 :
        return False
    else:
        for i in  range(2, int(n ** 0.5)+ 1) :
            if n % i == 0:
                return False
        
        return True
n = int(input())
print(prime_number_check(n))
```

# abi 4

```python
def perfect_number_check( n ) :
    sum = 0
    for i in  range(1, n + 1) :
        if n % i == 0 :
            sum += i
    if sum == n :
        return True
    else:
        return False
n = int(input())
print(perfect_number_check(n))

          
```

```python
def list_integer(n , k ) :
    count = 0
    length = 0
    for i in n :
        if n[count] == k :
            return count
        count += 1
        length += 1
    if count == length :
        return -1
n = list(map(int, input().split())) 

k = int(input())
print(list_integer(n , k))
```

# bai 6

```python
def giai_thua ( n) :
    tich = 1
    for i in range(1, n + 1) :
        tich *= i
    return tich
n = int(input())
print(giai_thua(n))
```
