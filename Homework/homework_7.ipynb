```python
# thuat toan thang dan
def selection_sort (n) :
    
    for i in range(len(n)) :
        min_index = i
        for j in range(i , len(n)) :
            if n[j] < n[min_index] :
                min_index = j
        n[i] , n[min_index] = n[min_index] , n[i]
    return n
# thuat toan giam dan
def insertion_sort(a) :
    for i in range(1,len(a)) :
        key = a[i]
        j = i -1 
        while (j >= 0 )  and key > a[j] :
            a[j+1] = a[j]
            j = j - 1
        a[j+ 1] = key
    return a
n = list(map(int, input().split())) 
lst_odd= []
lst_even = []
for i in range(len(n)) :
    if n[i] % 2== 0 :
        lst_odd.append(n[i])
    else:
        lst_even.append(n[i])
odd = selection_sort(lst_odd) 
even = insertion_sort(lst_even) 
result = odd + even
print(result)

       
           

    
```

```python

def combine(a,b) :
    is_ascending = True
    if len(a) > 1 :
        is_ascending = ( a[0] > a[-1])
    elif len(b) > 1 :
        is_ascending = (b[0] > b[-1])
    i , j = 0 , 0
    na , nb = len(a) , len(b)
    k = []
    while i < na :
        if is_ascending :
            if a[i] > b[j] :
                k.append(a[i])
                i += 1
                
            else:
                k.append(b[j])
                j += 1
        else:
            if a[i] < b[j] :
                k.append(a[i])
                i = i+ 1
            else:
                k.append(b[j]) 
                j = j + 1
    while i < na :
        k.append(a[i])
        i += 1
    while j < nb :
        k.append(b[j])
        j = j + 1
    return k

n = list(map(int, input().split())) 
m = list(map(int, input().split())) 
print(combine(n, m))
```
