# Working with Files in Python

This notebook demonstrates how to create files with data and work with them.

## Step 1: Create a file with your data

```python
# Create a file with data
data = """Name,Age,City,Job
John,25,New York,Engineer
Alice,30,Los Angeles,Doctor
Bob,28,Chicago,Teacher
Carol,35,Houston,Manager
David,26,Phoenix,Designer"""

# Write data to a file
with open('data.txt', 'w', encoding='utf-8') as f:
    f.write(data)

print("File 'data.txt' created successfully!")
```

## Step 2: Read the file and work with it

```python
# Read the file
with open('data.txt', 'r', encoding='utf-8') as f:
    content = f.read()

print("File content:")
print(content)
```

## Step 3: Parse and process the data

```python
# Using pandas to work with the file
import pandas as pd

# Read CSV file
df = pd.read_csv('data.txt')

print("DataFrame:")
print(df)
print("\nColumn names:", df.columns.tolist())
print("\nShape:", df.shape)
```

## Step 4: Filter and manipulate data

```python
# Example: Filter people older than 28
older_than_28 = df[df['Age'] > 28]
print("People older than 28:")
print(older_than_28)

# Example: Get only names and jobs
names_jobs = df[['Name', 'Job']]
print("\nNames and Jobs:")
print(names_jobs)
```

## Step 5: Save processed data to a new file

```python
# Save processed data to a new file
older_than_28.to_csv('data_filtered.txt', index=False)
print("Filtered data saved to 'data_filtered.txt'")

# Save to different formats
older_than_28.to_json('data_filtered.json', indent=2)
print("JSON format saved to 'data_filtered.json'")
```
