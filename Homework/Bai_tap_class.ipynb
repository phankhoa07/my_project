```python
class Cylinder :
    def __init__ (self, radius, height ) :
        self.r = radius
        self.h = height
    def get_surface_area ( self )  :
        pi = 3.14
        bottom_surface = pi * (self.r ** 2 ) * 2
        Lateral_surface_area = self.r * pi * self.h

        return  bottom_surface +Lateral_surface_area
        
    def get_volume ( self) :
        pi = 3.14
        volume = pi * self.r * self.r * self.h
        return volume
    def __str__(self):
        return f"Hình trụ có bán kính = {self.r} và chiều cao = {self.h}"

   
    def __repr__(self):
        return f"Cylinder(radius={self.r}, height={self.h})"
    
r , h = map(float,input().split())
c1 = Cylinder(r, h)
dien_tich = c1.get_surface_area()
the_tich = c1.get_volume()
print(f"dien tich cua hinh tru la {dien_tich:.2f}")
print(f"the tich cua hinh tru la {the_tich:.2f}")


```

```python
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h
        
    def scale(self, k):
        self.w *= k
        self.h *= k
        
    def area(self):
        return self.w * self.h
        
    def perimeter(self):
        return 2 * (self.w + self.h)

w, h, k = map(float, input().split())
rect = Rectangle(w, h)
rect.scale(k)
print(f"{rect.area():.2f} {rect.perimeter():.2f}")
```

```python
class Fraction:
    def __init__(self, num, den):
        self.x = num
        self.y = den

    def rut_gon(self):
        if self.x == 0:
            return "0"
            
        so_nho = min(abs(self.x), abs(self.y))
        
        for i in range(so_nho, 0, -1):
            if (self.x % i == 0) and (self.y % i == 0):
                self.x = self.x // i
                self.y = self.y // i
                break
                
        if self.y < 0:
            self.x = -self.x
            self.y = -self.y
            
        return f"{self.x}/{self.y}"

a, b, e, c, d = map(str, input().split())
a = int(a)
b = int(b)
c = int(c)
d = int(d)

if e == '+':
    tu  = a*d + c*b
    mau = b*d
    k = Fraction(tu, mau)
    print(k.rut_gon())
elif e == '-':
    tu = a*d - c*b
    mau = b*d
    k = Fraction(tu, mau)
    print(k.rut_gon())
elif e == '*':
    tu = a*c
    mau = b*d
    k = Fraction(tu, mau)
    print(k.rut_gon())
elif e == '/':
    tu = a*d
    mau = b*c
    k = Fraction(tu, mau)
    print(k.rut_gon())
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance
        
    def deposit(self, amount):
        self.balance += amount
        
    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            
    def __str__(self):
        return f"{self.owner}: {self.balance}"

data = input().split() 
owner_name = data[0]
initial_balance = float(data[1]) 

acc = BankAccount(owner_name, initial_balance)

n = int(input())

for _ in range(n):
    action_data = input().split()
    command = action_data[0]
    amount = float(action_data[1])
    
    if command == "DEPOSIT":
        acc.deposit(amount)
    elif command == "WITHDRAW":
        acc.withdraw(amount)

print(f"{acc.balance:.2f}")
```

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []

    def add_score(self, subject, score):
        self.scores.append(score)

    def avg(self):
        if len(self.scores) == 0:
            return 0.0
        return sum(self.scores) / len(self.scores)

    def rank(self):
        diem_tb = self.avg()
        if diem_tb >= 8.0:
            return "Excellent"
        elif diem_tb >= 6.5:
            return "Good"
        elif diem_tb >= 5.0:
            return "Average"
        else:
            return "Poor"

name = input().strip()
n = int(input().strip())
stu = Student(name)

for _ in range(n):
    data = input().split()
    score = float(data[-1])
    subject = " ".join(data[:-1])
    stu.add_score(subject, score)

print(f"{stu.name} {stu.avg():.2f} {stu.rank()}")
```

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

class Library:
    def __init__(self):
        self.books = []
        
    def add(self, book):
        self.books.append(book)
        
    def count_by_author(self, name):
        count = 0
        for b in self.books:
            if b.author == name:
                count += 1
        return count
        
    def find_by_year(self, y):
        count = 0
        for b in self.books:
            if b.year == y:
                count += 1
        return count

n = int(input().strip())
lib = Library()

for _ in range(n):
    line = input().strip().split(" ", 1)
    command = line[0]
    
    if command == "ADD":
        info = line[1].split(";")
        b = Book(info[0], info[1], info[2])
        lib.add(b)
    elif command == "COUNT":
        print(lib.count_by_author(line[1]))
    elif command == "COUNTYEAR":
        print(lib.find_by_year(line[1]))
```

```python
PI = 3.141592653589793

def custom_cos(x):
    term = 1.0
    cos_sum = 1.0
    x2 = x * x
    for i in range(1, 20):
        term *= -x2 / ((2 * i) * (2 * i - 1))
        cos_sum += term
    return cos_sum

def custom_acos(val):
    val = max(-1.0, min(1.0, val))
    low = 0.0
    high = PI
    for _ in range(70): 
        mid = (low + high) / 2.0
        if custom_cos(mid) > val:
            low = mid
        else:
            high = mid
    return (low + high) / 2.0

class Vector3D:
    def __init__(self, x, y, z):
        self.x = float(x)
        self.y = float(y)
        self.z = float(z)

    def __sub__(self, other):
        return Vector3D(self.x - other.x, self.y - other.y, self.z - other.z)

    def dot(self, other):
        return self.x * other.x + self.y * other.y + self.z * other.z

    def cross(self, other):
        return Vector3D(
            self.y * other.z - self.z * other.y,
            self.z * other.x - self.x * other.z,
            self.x * other.y - self.y * other.x
        )

    def magnitude(self):
        return (self.x**2 + self.y**2 + self.z**2) ** 0.5

class Plane:
    def __init__(self, p1, p2, p3):
        self.v1 = p2 - p1
        self.v2 = p3 - p1
        self.normal = self.v1.cross(self.v2)

    def get_angle_with(self, other_plane):
        n1 = self.normal
        n2 = other_plane.normal
        dot_prod = n1.dot(n2)
        if dot_prod < 0:
            dot_prod = -dot_prod
        mags = n1.magnitude() * n2.magnitude()
        if mags == 0:
            return 0.0
        cos_phi = dot_prod / mags
        phi_rad = custom_acos(cos_phi)
        phi_deg = phi_rad * (180.0 / PI)
        return phi_deg

def solve():
    data = []
    while len(data) < 12:
        try:
            line_input = input().split()
            data.extend(line_input)
        except EOFError:
            break
            
    if len(data) < 12:
        return

    A = Vector3D(data[0], data[1], data[2])
    B = Vector3D(data[3], data[4], data[5])
    C = Vector3D(data[6], data[7], data[8])
    D = Vector3D(data[9], data[10], data[11])

    plane_ABC = Plane(A, B, C)
    plane_BCD = Plane(B, C, D)

    angle = plane_ABC.get_angle_with(plane_BCD)
    print(f"{angle:.2f}")

if __name__ == '__main__':
    solve()
```
