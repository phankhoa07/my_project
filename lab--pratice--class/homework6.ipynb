# bai 16

```python
n = int(input()) # hang

m = int(input()) # cot

cot = []
for i in range(n):
    hang = []
    for j in range(m):
        num_in_row = int(input())
        hang.append(num_in_row)
    cot.append(hang)

for i in range(n):
    for j in range(m):
        chuoi_so = str(cot[i][j])
        do_dai = len(chuoi_so)
        
        so_khoang_trang = 4 - do_dai
        
        for k in range(so_khoang_trang):
            print(" ", end="")
            
        print(chuoi_so, end="")
    print()
```

# bai 17

```python
n = int(input()) # hang, #cot
duong_cheo_chinh = []
duong_cheo_phu = []



cot = []
for i in range(n):
    hang = []
    for j in range(n):
        num_in_row = int(input())
        hang.append(num_in_row)
    cot.append(hang)
for i in range(n) :
    so = cot[i][i]
    duong_cheo_chinh.append(so)
print(duong_cheo_chinh)
for i in range(n ) :
    so2 = cot[i][n - i - 1]
    duong_cheo_phu.append(so2)
print(duong_cheo_phu)






     
```

# bai 18


```python
n, m , k = map(int,input().split())

cot = []
for i in range(n):
    hang = []
    for j in range(m):
        num_in_row = int(input())
        hang.append(num_in_row)
    cot.append(hang)
i,j = 0, 0
# tinh tong cac so tren hang
tong_hang = 0 
while j < m :
    tong_hang += cot[k][j]
    j = j + 1
print(tong_hang)
# tinh cac so tren cot
tong_cot = 0
while i < n :
    tong_cot += cot[i][k]
    i = i + 1
print(tong_cot)


```
